##**5. Implement a Python program to compute the mean, median, and mode of a given list of numbers.**
**numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]**

```python
import statistics as stats

numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]

mean_value = stats.mean(numbers)
median_value = stats.median(numbers)
mode_value = stats.mode(numbers)


print("Numbers:", numbers)
print("Mean:", mean_value)
print("Median:", median_value)
print("Mode:", mode_value)
```

##**6. Compute the covariance and correlation coefficient between the following two datasets provided as lists in Python:**
**list_x = [10, 20, 30, 40, 50],
list_y = [15, 25, 35, 45, 60]**

```python
import numpy as np

list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]

x = np.array(list_x)
y = np.array(list_y)

cov_matrix = np.cov(x, y, bias=False)  # bias=False → sample covariance
covariance = cov_matrix[0, 1]

correlation = np.corrcoef(x, y)[0, 1]

print("List X:", list_x)
print("List Y:", list_y)
print("Covariance:", covariance)
print("Correlation Coefficient:", correlation)
```

##**7. Write a Python script to draw a boxplot for the following numeric list and identify its outliers. Explain the result:**
**data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]**

```python
import matplotlib.pyplot as plt
import numpy as np

data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]

plt.boxplot(data, vert=False, patch_artist=True, boxprops=dict(facecolor="lightblue"))
plt.title("Boxplot of Given Data")
plt.xlabel("Values")
plt.show()

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR


outliers = [x for x in data if x < lower_bound or x > upper_bound]


print(f"Q1 (25th percentile): {Q1}")
print(f"Q3 (75th percentile): {Q3}")
print(f"IQR (Q3 - Q1): {IQR}")
print(f"Lower Bound: {lower_bound}")
print(f"Upper Bound: {upper_bound}")
print(f"Outliers: {outliers}")
```

##**8. You are working as a data analyst in an e-commerce company. The marketing team wants to know if there is a relationship between advertising spend and daily sales.                a) Explain how you would use covariance and correlation to explore this relationship.                         b) Write Python code to compute the correlation between the two lists:**
**advertising_spend = [200, 250, 300, 400, 500],  
daily_sales = [2200, 2450, 2750, 3200, 4000]**

###ANSWERS
**a)Covariance**

Definition:-
Covariance measures how two variables change together.
If both increase or decrease simultaneously → positive covariance.
If one increases while the other decreases → negative covariance.
If no consistent pattern → covariance ≈ 0.

Limitation:-
Covariance only shows direction of the relationship, not strength, and depends on the scale of units (e.g., dollars, sales).

**Correlation**

Definition:-
Correlation is a standardized measure of covariance that quantifies both strength and direction of a linear relationship.

Range:-
r∈[−1,+1]


r=+1: Perfect positive linear relationship

r=−1: Perfect negative linear relationship

r=0: No linear relationship

**Interpretation:-**

If covariance > 0 and correlation ≈ +1 → higher ad spend leads to higher daily sales.

If covariance < 0 → inverse relationship (not likely here).

If correlation ≈ 0 → no linear relationship between ad spend and sales.

So, both metrics help you determine whether marketing investments are driving sales.

```python
#b)
import numpy as np

advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]

x = np.array(advertising_spend)
y = np.array(daily_sales)

cov_matrix = np.cov(x, y, bias=False)
covariance = cov_matrix[0, 1]

correlation = np.corrcoef(x, y)[0, 1]

print("Advertising Spend:", x)
print("Daily Sales:", y)
print(f"Covariance: {covariance:.2f}")
print(f"Correlation Coefficient: {correlation:.3f}")
```

##**9. Your team has collected customer satisfaction survey data on a scale of 1-10 and wants to understand its distribution before launching a new product. a)Explain which summary statistics and visualizations (e.g. mean, standard deviation, histogram) you’d use.b) Write Python code to create a histogram using Matplotlib for the survey data:**
**survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]**

###ANSWERS
**(a) Explanation**

To understand the distribution of customer satisfaction scores, I would begin by calculating key summary statistics. The mean provides the average satisfaction level, showing the overall trend of how customers feel. The median identifies the middle score, which is especially useful if the data is skewed or contains outliers. The mode highlights the most frequently occurring satisfaction score, giving insight into the most common rating customers gave. I’d also calculate the standard deviation, which measures how spread out the scores are — a low standard deviation would mean most customers gave similar ratings, while a high one would suggest a wide variation in opinions. Additionally, the minimum and maximum values help identify the range of responses, showing how low or high satisfaction levels went.

For visual analysis, I would use a **histogram** to display how frequently each satisfaction score appears. This makes it easy to spot whether the data is roughly symmetric, right-skewed, or left-skewed. A **boxplot** could also be useful to show the median, quartiles, and any potential outliers. These visual tools, together with the numerical summaries, provide a clear understanding of the overall customer satisfaction distribution — helping the team see if most customers are satisfied and whether there’s any significant variation in their responses.

```python
import matplotlib.pyplot as plt
import numpy as np

survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]

mean_value = np.mean(survey_scores)
median_value = np.median(survey_scores)
std_dev = np.std(survey_scores, ddof=1)

print("Mean:", round(mean_value, 2))
print("Median:", median_value)
print("Standard Deviation:", round(std_dev, 2))


plt.hist(survey_scores, bins=6, color='lightblue', edgecolor='black')
plt.title("Customer Satisfaction Distribution")
plt.xlabel("Satisfaction Score (1-10)")
plt.ylabel("Frequency")
```
